# **Range**
In Python, range() is a built-in function used to generate a sequence of numbers. It's often used in for loops.

Example: range(5) generates 0, 1, 2, 3, 4.

### Using `range()` in a loop

````python
# Example 1: Loop from 0 to 4
print("Counting from 0 to 4:")
for i in range(5):
    print(i)

# Example 2: Loop from 2 to 6
print("\nCounting from 2 to 6:")
for j in range(2, 7):
    print(j)

# Example 3: Loop from 1 to 9, stepping by 2
print("\nCounting from 1 to 9, step 2:")
for k in range(1, 10, 2):
    print(k)
````

````python
print(list(range(10)))
print(list(range(2,10)))
print(list(range(2,10,2)))
````

# **Python Loops,Function and Modules**

### Loops

Loops are used to execute a block of code repeatedly. Python provides two main types of loops:

1.  **`for` loop**: Used for iterating over a sequence (like a list, tuple, dictionary, set, or string) or other iterable objects. The `for` loop executes a block of code for each item in the sequence.

    **Example:**
    ```python
    fruits = ["apple", "banana", "cherry"]
    for x in fruits:
        print(x)
    ```

2.  **`while` loop**: Used to execute a block of code as long as a specified condition is true. It continues to loop as long as the condition remains `True`.

    **Example:**
    ```python
    i = 1
    while i < 6:
        print(i)
        i += 1
    ```

**Loop Control Statements:**
*   **`break`**: Terminates the loop entirely.
*   **`continue`**: Skips the current iteration and moves to the next.
*   **`pass`**: A null operation; nothing happens when it executes. It can be used as a placeholder where a statement is syntactically required but you don't want any code to execute.

### Functions

A function is a block of organized, reusable code that performs a single, related action. Functions provide better modularity for your application and a high degree of code reusing.

**Key concepts:**
*   **Defining a function**: Use the `def` keyword.
*   **Calling a function**: Use the function name followed by parentheses `()`.
*   **Parameters/Arguments**: Information can be passed into functions as arguments.
*   **`return` statement**: Used to send a result back to the caller.

**Example:**
```python
def greet(name):
    """This function greets the person passed in as a parameter"""
    return "Hello, " + name + "!"

# Call the function
message = greet("Alice")
print(message)
```

**Types of functions:**
*   **Built-in functions**: Functions that are pre-defined in Python (e.g., `print()`, `len()`, `type()`).
*   **User-defined functions**: Functions created by the user to perform specific tasks.
*   **Lambda functions**: Small, anonymous functions defined using the `lambda` keyword. They can take any number of arguments but can only have one expression.

### Modules

A module is simply a file containing Python code (variables, functions, classes, etc.). Modules help to organize code logically and make it reusable. When you break down code into modules, each module can focus on a specific task, making the overall project easier to understand and maintain.

**Key concepts:**
*   **Creating a module**: Save your Python code in a `.py` file.
*   **Importing a module**: Use the `import` statement to make the contents of a module available in another script or module.

**Example:**

*   **`my_module.py` (file content):**
    ```python
    def add(a, b):
        return a + b

    greeting = "Hello from my_module!"
    ```

*   **In another Python script/notebook:**
    ```python
    import my_module

    result = my_module.add(10, 5)
    print(result) # Output: 15
    print(my_module.greeting) # Output: Hello from my_module!
    ```

**Ways to import:**
*   `import module_name`: Imports the entire module.
*   `import module_name as alias`: Imports the module and gives it an alias.
*   `from module_name import item_name`: Imports specific items (functions, variables, classes) from the module.
*   `from module_name import *`: Imports all items from the module (generally discouraged as it can lead to name clashes).
