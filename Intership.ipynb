```python
a=23
b=54
is_true = True
is_false = False
print(f"Boolean True: {is_true}, Type: {type(is_true)}")
print(f"Boolean False: {is_false}, Type: {type(is_false)}")
print(type(a))
binary_data = b'\x00\x01\x02ABC'
print(f"Binary Data: {binary_data}, Type: {type(binary_data)}")
```
