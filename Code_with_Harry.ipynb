# **Match Case**

the match case statement introduced in python 3.10 provides pattern matching similar to switch statements in other languages .





**Syntax**

```python
def http_status (code):
  match code :
    case 200:
      return "ok"
    case 400:
      return "Bad Request"
    case 404:
      return "Not Found "
    case 500:
      return "Internal Server Error"
    case _:
      return "Unknown Status"

print(http_status(200)) # Output : OK
print(http_status(404)) # Output: Not Found
```

```python
code = 200
match code:
    case 200:
        print("OK")
    case 400:
        print("Bad Request")
    case 404:
        print("Not Found")
    case 500:
        print("Internal Server Error")
    case _:
        print("Unknown Status")
```

**3.Loop Control Statements**

a. break : Exits the loop.

b. Continue: Skips to the iteration.

c. pass: Does nothing (used as a placeholder ).


```python
for i in range (5):
  if i == 3:
    break
  print(i)
```

```python
for i in range (5):
  if i == 3:
    continue
  print(i)
```

```python
for i in range (5):
  if i == 3:
    pass
  print(i)
```

# **Dictionary and Dictionary Methods**

A dictionary in Python is an unordered, mutable, and key-value pair collection. It allows efficient data retrieval and modification. Dictionaries in Python are ordered as of Python 3.7 .




```python
# Empty dictionary
empty_dict = {}

# Dictionary with key-value pairs
student = {
    "name": "Alice",
    "age": 25,
    "grade": "A"
}

# Using dict() constructor
person = dict(name="John", age=30, city="New York")

print(student)
print(person)
```

| **Method** | **Description** | **Example** |
|---|---|---|
| `keys()` | Returns all keys in the dictionary. | `student.keys()` |
| `values()` | Returns all values in the dictionary. | `student.values()` |
| `items()` | Returns all key-value pairs as tuples. | `student.items()` |
| `get(key, default)` | Returns the value for `key`, or `default` if the key is not found. | `student.get("age", 0)` |
| `update(dict2)` | Merges `dict2` into the dictionary. | `student.update({"age": 26})` |
| `pop(key, default)` | Removes the specified key and returns its value, or `default` if the key is not found. | `student.pop("grade")` |
| `popitem()` | Removes and returns the **last inserted** key-value pair. | `student.popitem()` |
| `setdefault(key, default)` | Returns the value for `key`; if not found, inserts the key with the given default value. | `student.setdefault("city", "Unknown")` |
| `clear()` | Removes all items from the dictionary. | `student.clear()` |
| `copy()` | Returns a shallow copy of the dictionary. | `new_dict = student.copy()` |

```python
# Dictionary with key-value pairs
student = {
    "name": "Alice",
    "age": 25,
    "grade": "A"
}

print(student["age"])
student["age"]=26
print(student["age"])

print(student)

print(student.keys())
print(student.values())
print(student.items())
student.update({"name":"Aniket", "language":"English"})
print(student)
```

# **File handling**

File handling allows Python programs to read, write, and manipulate files stored on disk. Python provides built-in functions for working with files

```python
s="harry is a good boy and he teaches us english ."

f=open("sample.txt","w")
f.write(s)
f.close()

# f=open("sample.txt","r")
# print(f.read())
# f.close()

with open("sample.txt","r") as f:
  content=f.read()
print(content)
```

| **Mode** | **Name** | **Description** | **Example** |

| `'w'` | Write | Opens a file for writing. Creates a new file if it doesn't exist and overwrites existing content. | `open("file.txt", "w")` |
| `'a'` | Append | Opens a file for writing. Creates a new file if it doesn't exist and adds content at the end without overwriting. | `open("file.txt", "a")` |

| `'b'` | Binary | Used for working with binary/non-text files such as images and PDFs. | `open("image.jpg", "rb")` |
| `'t'` | Text | Text mode (default), used for working with text files. | `open("file.txt", "rt")` |

# **JSON Module**

JSON (JavaScript Object Notation) is a lightweight data format used for data exchange between servers and applications. It is widely used in APIs, web applications, and configurations

```python
import json

object={"Name":"Alice","Age":25,"City":"New York"}
json_data=json.dumps(object)
print(json_data)
print(type(json_data))


d=json.loads(json_data)
print(d)
print(type(d))


c=json.dumps(d,indent=4)
print(c)


file=open("data.json","w")
json.dump(d,file,indent=4)
file.close()
```
