# **String Methods**

| #  | Method           | What it does                                    | Example                      | Output             |

| 4  | `upper()`        | Converts to uppercase                           | `"hello".upper()`            | `HELLO`            |

# **Operators in Python**

**1.Arithmatic Operator**

| Operator | Name           | Example   |     Result |
| -------- | -------------- | --------- | ---------: |
| `+`      | Addition       | `10 + 3`  |       `13` |
| `-`      | Subtraction    | `10 - 3`  |        `7` |
| `*`      | Multiplication | `10 * 3`  |       `30` |
| `/`      | Division       | `10 / 3`  | `3.333...` |
| `//`     | Floor Division | `10 // 3` |        `3` |
| `%`      | Modulus        | `10 % 3`  |        `1` |
| `**`     | Exponent       | `10 ** 3` |     `1000` |


```python
a=45
b=6
print(a+b)

print(a-b)

print(a*b)

print(a/b)

print(a//b)

print(a%b)

print(a**b)
```

2.Coparision / Relational Operator


| Operator | Meaning               | Example  | Result  |
| -------- | --------------------- | -------- | ------- |
| `==`     | Equal                 | `5 == 5` | `True`  |
| `!=`     | Not Equal             | `5 != 3` | `True`  |
| `>`      | Greater than          | `5 > 3`  | `True`  |
| `<`      | Less than             | `5 < 3`  | `False` |
| `>=`     | Greater than or equal | `5 >= 5` | `True`  |
| `<=`     | Less than or equal    | `5 <= 3` | `False` |


```python
c=34
d=45
print(c==d)

print(c!=d)

print(c>d)

print(c<d)

print(c>=d)

print(c<=d)
```

```python
a=6
b=6
print(a==b)
print(a!=b)
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
```

| Operator | Example   | Equivalent to |      |        |    |
| -------- | --------- | ------------- | ---- | ------ | -- |
| `=`      | `x = 10`  | `x = 10`      |      |        |    |
| `+=`     | `x += 5`  | `x = x + 5`   |      |        |    |
| `-=`     | `x -= 5`  | `x = x - 5`   |      |        |    |
| `*=`     | `x *= 5`  | `x = x * 5`   |      |        |    |
| `/=`     | `x /= 5`  | `x = x / 5`   |      |        |    |
| `//=`    | `x //= 5` | `x = x // 5`  |      |        |    |
| `%=`     | `x %= 5`  | `x = x % 5`   |      |        |    |
| `**=`    | `x **= 5` | `x = x ** 5`  |      |        |    |
| `&=`     | `x &= 5`  | `x = x & 5`   |      |        |    |
| `^=`     | `x ^= 5`  | `x = x ^ 5`   |      |        |    |
| `>>=`    | `x >>= 2` | `x = x >> 2`  |      |        |    |
| `<<=`    | `x <<= 2` | `x = x << 2`  |      |        |    |


**4.Logical Operator**
| Operator | Meaning              | Example            | Result  |
| -------- | -------------------- | ------------------ | ------- |
| `and`    | Both conditions True | `5 > 2 and 10 > 5` | `True`  |
| `or`     | At least one True    | `5 > 10 or 10 > 5` | `True`  |
| `not`    | Reverses result      | `not(5 > 2)`       | `False` |




```python
age = 20

print(age >= 18 and age <= 60)
```

**5.Bitwise Operator**



| Operator | Name        | Example  |    |    |
| -------- | ----------- | -------- | -- | -- |
| `&`      | AND         | `5 & 3`  |    |    |
|
| `^`      | XOR         | `5 ^ 3`  |    |    |
| `~`      | NOT         | `~5`     |    |    |
| `<<`     | Left Shift  | `5 << 1` |    |    |
| `>>`     | Right Shift | `5 >> 1` |    |    |


```python
a = 5      # 101
b = 3      # 011

print(a & b)
```

**6.Membership Operator**

| Operator | Meaning     | Example              | Result |
| -------- | ----------- | -------------------- | ------ |
| `in`     | Present     | `"a" in "apple"`     | `True` |
| `not in` | Not present | `"z" not in "apple"` | `True` |


```python
name = "Python"

print("P" in name)
```
