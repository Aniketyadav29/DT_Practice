# **Operator Precedence**

Operator precedence determines the order in which Python evaluates operators in an expression.

**Precedence Table — Highest to Lowest**

| Priority | Operator                         | Description              | Example          |
| -------: | -------------------------------- | ------------------------ | ---------------- |
|        1 | `()`                             | Parentheses              | `(2 + 3) * 4`    |
|        2 | `**`                             | Exponentiation           | `2 ** 3`         |
|        3 | `+x`, `-x`, `~x`                 | Unary operators          | `-5`             |
|        4 | `*`, `/`, `//`, `%`              | Multiplication, Division | `10 * 2`         |
|        5 | `+`, `-`                         | Addition, Subtraction    | `10 + 2`         |
|        6 | `<<`, `>>`                       | Bitwise shifts           | `8 >> 2`         |
|        7 | `&`                              | Bitwise AND              | `5 & 3`          |
|        8 | `^`                              | Bitwise XOR              | `5 ^ 3`          |
|        9 | `\|`                             | Bitwise OR               | `5 \| 3`         |
|       10 | `==`, `!=`, `>`, `<`, `>=`, `<=` | Comparisons              | `5 > 3`          |
|       11 | `is`, `is not`                   | Identity                 | `a is b`         |
|       12 | `in`, `not in`                   | Membership               | `x in list`      |
|       13 | `not`                            | Logical NOT              | `not True`       |
|       14 | `and`                            | Logical AND              | `True and False` |
|       15 | `or`                             | Logical OR               | `True or False`  |
|       16 | `=`, `+=`, `-=`, `*=`, etc.      | Assignment               | `x += 5`         |

# **Match Case**

the match case statement introduced in python 3.10 provides pattern matching similar to switch statements in other languages .





**Syntax**

```python
def http_status (code):
  match code :
    case 200:
      return "ok"
    case 400:
      return "Bad Request"
    case 404:
      return "Not Found "
    case 500:
      return "Internal Server Error"
    case _:
      return "Unknown Status"

print(http_status(200)) # Output : OK
print(http_status(404)) # Output: Not Found
```

```python
code = 200
match code:
    case 200:
        print("OK")
    case 400:
        print("Bad Request")
    case 404:
        print("Not Found")
    case 500:
        print("Internal Server Error")
    case _:
        print("Unknown Status")
```
