# **String Methods**

| #  | Method           | What it does                                    | Example                      | Output             |
| -- | ---------------- | ----------------------------------------------- | ---------------------------- | ------------------ |
| 1  | `capitalize()`   | First character uppercase                       | `"hello".capitalize()`       | `Hello`            |
| 2  | `casefold()`     | Converts to lowercase (stronger than `lower()`) | `"HELLO".casefold()`         | `hello`            |
| 3  | `lower()`        | Converts to lowercase                           | `"HELLO".lower()`            | `hello`            |
| 4  | `upper()`        | Converts to uppercase                           | `"hello".upper()`            | `HELLO`            |
| 5  | `title()`        | First letter of each word uppercase             | `"hello world".title()`      | `Hello World`      |
| 6  | `swapcase()`     | Uppercase → lowercase & vice versa              | `"Hello".swapcase()`         | `hELLO`            |
| 7  | `center()`       | Centers the string                              | `"Hi".center(6, "-")`        | `--Hi--`           |
| 8  | `count()`        | Counts occurrences                              | `"banana".count("a")`        | `3`                |
| 9  | `find()`         | Returns first position of substring             | `"hello".find("l")`          | `2`                |
| 10 | `rfind()`        | Finds substring from right                      | `"hello".rfind("l")`         | `3`                |
| 11 | `index()`        | Finds position; error if not found              | `"hello".index("e")`         | `1`                |
| 12 | `rindex()`       | Finds position from right                       | `"hello".rindex("l")`        | `3`                |
| 13 | `startswith()`   | Checks starting characters                      | `"hello".startswith("he")`   | `True`             |
| 14 | `endswith()`     | Checks ending characters                        | `"hello".endswith("lo")`     | `True`             |
| 15 | `replace()`      | Replaces substring                              | `"hello".replace("h","y")`   | `yello`            |
| 16 | `strip()`        | Removes spaces from both sides                  | `" hello ".strip()`          | `hello`            |
| 17 | `lstrip()`       | Removes left-side spaces                        | `" hello".lstrip()`          | `hello`            |
| 18 | `rstrip()`       | Removes right-side spaces                       | `"hello ".rstrip()`          | `hello`            |
| 19 | `split()`        | Converts string into list                       | `"a,b,c".split(",")`         | `['a','b','c']`    |
| 20 | `rsplit()`       | Splits from right                               | `"a-b-c".rsplit("-",1)`      | `['a-b','c']`      |
| 21 | `splitlines()`   | Splits at line boundaries                       | `"a\nb".splitlines()`        | `['a','b']`        |
| 22 | `join()`         | Joins elements into string                      | `"-".join(["a","b"])`        | `a-b`              |
| 23 | `partition()`    | Splits into 3 parts                             | `"a-b".partition("-")`       | `('a','-','b')`    |
| 24 | `rpartition()`   | Partition from right                            | `"a-b-c".rpartition("-")`    | `('a-b','-','c')`  |
| 25 | `zfill()`        | Adds zeros to left                              | `"42".zfill(5)`              | `00042`            |
| 26 | `expandtabs()`   | Replaces tabs with spaces                       | `"Hi\tWorld".expandtabs(4)`  | `Hi  World`        |
| 27 | `translate()`    | Replaces characters using mapping               | `"abc".translate(...)`       | Depends on mapping |
| 28 | `maketrans()`    | Creates translation table                       | `str.maketrans("abc","123")` | Mapping table      |
| 29 | `isalpha()`      | Checks only alphabets                           | `"Hello".isalpha()`          | `True`             |
| 30 | `isalnum()`      | Checks alphabets + numbers                      | `"Hello123".isalnum()`       | `True`             |
| 31 | `isdecimal()`    | Checks decimal characters                       | `"123".isdecimal()`          | `True`             |
| 32 | `isdigit()`      | Checks digits                                   | `"123".isdigit()`            | `True`             |
| 33 | `isnumeric()`    | Checks numeric characters                       | `"123".isnumeric()`          | `True`             |
| 34 | `isidentifier()` | Checks valid Python identifier                  | `"my_var".isidentifier()`    | `True`             |
| 35 | `islower()`      | Checks all cased chars lowercase                | `"hello".islower()`          | `True`             |
| 36 | `isupper()`      | Checks all cased chars uppercase                | `"HELLO".isupper()`          | `True`             |
| 37 | `istitle()`      | Checks title case                               | `"Hello World".istitle()`    | `True`             |
| 38 | `isspace()`      | Checks whitespace characters                    | `"   ".isspace()`            | `True`             |
| 39 | `isascii()`      | Checks ASCII characters                         | `"Hello".isascii()`          | `True`             |
| 40 | `isprintable()`  | Checks printable characters                     | `"Hello".isprintable()`      | `True`             |
| 41 | `isdecimal()`    | Checks decimal digits                           | `"123".isdecimal()`          | `True`             |
