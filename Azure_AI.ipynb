# **3.Batch Processing:**

Batch Processing is a method of processing data where a large amount of data is collected and processed together at a scheduled time, instead of processing each piece of data immediately.

Example:
A company collects all employees' attendance data throughout the day and processes it together at the end of the day to calculate working hours.

Key Points:

Data is collected in batches.
Processing happens at a specific/scheduled time.
Suitable for large amounts of data.
It does not require real-time processing.

Simple example:
📥 Collect data → 📦 Make a batch → ⚙️ Process together → 📊 Get result

# **4.Streaming Data**

Streaming Data is a method of processing data where data is processed continuously and immediately as it is generated or received, instead of waiting for a large amount of data to accumulate.

Example:
A food delivery app continuously receives the driver's GPS location and updates the customer's map in real time.

Key Points:

Data is processed continuously.
Processing happens in real time or near real time.
Suitable for continuously generated data.
Useful when quick results or immediate action is required.

Examples:

📍 GPS location tracking
💳 Online payment transactions
📱 Social media feeds
📈 Stock market prices
🌡️ IoT sensor data

Simple flow:
Data Generated → Process Immediately → Result

| Feature             | **Batch Processing**                                   | **Streaming Data Processing**                  |
| ------------------- | ------------------------------------------------------ | ---------------------------------------------- |
| **Meaning**         | Processes data in **groups/batches**                   | Processes data **continuously as it arrives**  |
| **Processing time** | At a **scheduled time**                                | **Immediately or near real-time**              |
| **Data size**       | Usually handles **large volumes** together             | Handles **continuous flow of data**            |
| **Result**          | Results are available **after the batch is processed** | Results are available **almost immediately**   |

| **Cost/complexity** | Generally **simpler** to implement                     | Generally **more complex** to implement        |

# **Introduction of Databases**

1. Database

A database is an organized collection of data that allows us to store, manage, search, and retrieve information efficiently.

Example: A college database stores student names, roll numbers, courses, marks, etc.

2. DBMS (Database Management System)

A DBMS is software used to create, store, update, delete, and manage data in a database.

Examples: MySQL, PostgreSQL, Oracle, SQL Server.

3. Table

A table stores data in the form of rows and columns.

Row → One complete record
Column → One attribute/property

Example: Student(ID, Name, Course, Marks)

4. SQL

SQL (Structured Query Language) is used to communicate with and manage relational databases.

Common operations:
SELECT, INSERT, UPDATE, DELETE

5. Primary Key

A primary key uniquely identifies each record in a table.

Example: Student_ID

6. Foreign Key

A foreign key connects one table with another table and maintains relationships between them.

7. Database Schema

A schema defines the structure of a database—tables, columns, relationships, constraints, etc.

8. Data Integrity

Data integrity means keeping data accurate, consistent, and reliable.

9. Database Security

It protects data from unauthorized access, modification, or deletion using authentication, permissions, and access control.

10. Transactions

A transaction is a group of database operations treated as one unit.

Example: In online banking, money is deducted from one account and added to another as a single transaction.

# **Basics Of SQL Quiries :**

SQL (Structured Query Language) is the standard language used to communicate with and manage relational databases. It allows you to perform various operations, from retrieving data to modifying database structures. Here are some fundamental SQL concepts and commands:

### 1. Data Definition Language (DDL)
DDL commands are used to define, modify, or delete database structures.

*   **`CREATE`**: Used to create databases or tables.
    ```sql
    -- Create a new database
    CREATE DATABASE MyDatabase;

    -- Create a new table named 'Employees'
    CREATE TABLE Employees (
        EmployeeID INT PRIMARY KEY,
        FirstName VARCHAR(50),
        LastName VARCHAR(50),
        Department VARCHAR(50)
    );
    ```

*   **`ALTER`**: Used to modify the structure of an existing table.
    ```sql
    -- Add a new column 'Email' to the Employees table
    ALTER TABLE Employees
    ADD Email VARCHAR(100);

    -- Modify the data type of the 'Department' column
    ALTER TABLE Employees
    ALTER COLUMN Department VARCHAR(75);
    ```

*   **`DROP`**: Used to delete an existing database or table.
    ```sql
    -- Delete the Employees table
    DROP TABLE Employees;

    -- Delete the MyDatabase database
    DROP DATABASE MyDatabase;
    ```

### 2. Data Manipulation Language (DML)
DML commands are used to manage data within database objects.

*   **`SELECT`**: Used to retrieve data from one or more tables. This is the most frequently used SQL command.
    ```sql
    -- Select all columns and all rows from the Employees table
    SELECT * FROM Employees;

    -- Select specific columns (FirstName, LastName) from the Employees table
    SELECT FirstName, LastName FROM Employees;

    -- Select employees from the 'Sales' department
    SELECT * FROM Employees
    WHERE Department = 'Sales';
    ```

*   **`INSERT INTO`**: Used to add new rows of data into a table.
    ```sql
    -- Insert a new employee record
    INSERT INTO Employees (EmployeeID, FirstName, LastName, Department)
    VALUES (1, 'John', 'Doe', 'Sales');

    -- Insert another employee, providing values for all columns implicitly
    INSERT INTO Employees
    VALUES (2, 'Jane', 'Smith', 'Marketing', 'jane.smith@example.com');
    ```

*   **`UPDATE`**: Used to modify existing data in a table.
    ```sql
    -- Update the department for EmployeeID 1 to 'Marketing'
    UPDATE Employees
    SET Department = 'Marketing'
    WHERE EmployeeID = 1;
    ```

*   **`DELETE FROM`**: Used to remove existing rows of data from a table.
    ```sql
    -- Delete the employee with EmployeeID 2
    DELETE FROM Employees
    WHERE EmployeeID = 2;

    -- Delete all records from the Employees table (use with caution!)
    -- DELETE FROM Employees;
    ```

### 3. Basic Clauses

*   **`WHERE`**: Filters records based on a specified condition.
    ```sql
    SELECT FirstName, LastName FROM Employees
    WHERE Department = 'Sales' AND EmployeeID > 5;
    ```

*   **`ORDER BY`**: Sorts the result set by one or more columns.
    ```sql
    SELECT * FROM Employees
    ORDER BY LastName ASC, FirstName DESC; -- Sort by last name ascending, then first name descending
    ```

*   **`GROUP BY`**: Groups rows that have the same values in specified columns into a summary row. Often used with aggregate functions (e.g., `COUNT`, `SUM`, `AVG`, `MAX`, `MIN`).
    ```sql
    -- Count the number of employees in each department
    SELECT Department, COUNT(EmployeeID) AS NumberOfEmployees
    FROM Employees
    GROUP BY Department;
    ```

*   **`HAVING`**: Filters groups based on a specified condition, used after `GROUP BY`.
    ```sql
    -- Find departments with more than 5 employees
    SELECT Department, COUNT(EmployeeID) AS NumberOfEmployees
    FROM Employees
    GROUP BY Department
    HAVING COUNT(EmployeeID) > 5;
    ```

### 4. Joins
Joins are used to combine rows from two or more tables based on a related column between them.

*   **`INNER JOIN`**: Returns rows when there is a match in both tables.
    ```sql
    SELECT E.FirstName, E.LastName, D.DepartmentName
    FROM Employees AS E
    INNER JOIN Departments AS D
    ON E.DepartmentID = D.DepartmentID;
    ```

*   **`LEFT (OUTER) JOIN`**: Returns all rows from the left table, and the matching rows from the right table. If no match, NULL is used for the right table's columns.
    ```sql
    SELECT E.FirstName, E.LastName, D.DepartmentName
    FROM Employees AS E
    LEFT JOIN Departments AS D
    ON E.DepartmentID = D.DepartmentID;
    ```

*   **`RIGHT (OUTER) JOIN`**: Returns all rows from the right table, and the matching rows from the left table. If no match, NULL is used for the left table's columns.

*   **`FULL (OUTER) JOIN`**: Returns all rows when there is a match in one of the tables.

This covers the most basic and frequently used SQL commands and concepts. Mastering these will provide a strong foundation for working with relational databases.

# **SQL Server on Azure Virtual Machines**

**Key Concepts:**

*   **Complete operating system control**: Users have full control over the underlying operating system.
*   **Install custom software**: Allows installation of custom software, including specific SQL Server features or third-party applications.
*   **Infrastructure as a Service (IaaS)**: Provides virtualized computing resources over the internet.

**Explanation:**

*   Suitable when applications require full SQL Server features that might not be available in Azure SQL Database (PaaS offerings).
*   Offers greater control and customization for specific application requirements or legacy systems.
*   Requires more management overhead compared to Azure SQL Database, as users are responsible for OS patching, backups, and high availability configurations.

### **When to choose SQL Server on Azure Virtual Machines:**

Choose SQL Server on Azure VMs when you need:

*   **Full control over the SQL Server environment**: This includes the operating system and SQL Server configuration.
*   **Support for specific SQL Server features**: If your application relies on features not available in Azure SQL Database (PaaS).
*   **Lift-and-shift of on-premises applications**: For easy migration of existing SQL Server workloads with minimal code changes.
*   **Customization**: When you need to install other applications on the same server or have complex configurations.

# **Azure SQL  Vs Managed Instance**



Azure SQL Managed Instance is a fully managed, intelligent, and scalable cloud database service that combines the broadest SQL Server engine compatibility with all the benefits of a fully managed PaaS (Platform as a Service). It's designed for migrations of on-premises applications to the cloud with minimal changes to the application.

### **Key Concepts:**

*   **Near 100% SQL Server compatibility**: Allows for easy lift-and-shift of existing SQL Server applications to Azure with virtually no code changes.
*   **Fully managed PaaS service**: Microsoft handles infrastructure management, patching, backups, and updates, reducing operational overhead.
*   **Instance-scoped features**: Supports SQL Server features at the instance level (e.g., SQL Server Agent, Common Language Runtime (CLR), cross-database queries) that are not available in Azure SQL Database (single database).
*   **Isolation and security**: Deployed within a dedicated virtual network (VNet) in Azure, providing network isolation and strong security.
*   **Scalability**: Compute and storage can be scaled independently, allowing resources to be adjusted based on workload demands.
*   **Built-in High Availability and Disaster Recovery**: Includes features like automated backups and geo-replication to ensure data availability and protection.
*   **Developer-friendly**: Supports various programming languages and tools, similar to SQL Server on Azure VM and Azure SQL Database.

### **When to choose Azure SQL Managed Instance:**

Choose Azure SQL Managed Instance when you need:

*   **To migrate existing SQL Server applications**: Especially those that require instance-scoped features or have specific network isolation requirements.
*   **A fully managed database service with broad SQL Server compatibility**: It bridges the gap between Azure SQL Database (PaaS) and SQL Server on Azure VMs (IaaS).
*   **To modernize applications**: Without having to refactor the database layer significantly.
*   **Predictable performance and cost**: With the flexibility of cloud scaling and managed service benefits.
