**1.Why Do Organization Need Data Solutions ?**

Modern organizations continually generate data spanning transactions, financials, devices, networks, and business operations.

Data solutions provide ways to collect, store, organize, process, and analyze that data.

Organizations use data to serve business needs, including performance, credit decisions, and customer services.

Without suitable data solutions, data can become difficult to access, understand, operate in, maintain, or analyze

**2. Types of Data**

**Structured**

Fixed organization, easily stored and defined.
Examples: employee data, sales transactions, and records.

Easily stored with defined database tools.

**Semi-Structured / Unstructured :**

Some structured (JSON, XML, log files), and unstructured content, videos, audio, PDFs, etc.

Difficult to store and process using traditional approaches.

**3.Batch Processing:**

Batch Processing is a method of processing data where a large amount of data is collected and processed together at a scheduled time, instead of processing each piece of data immediately.

Example:
A company collects all employees' attendance data throughout the day and processes it together at the end of the day to calculate working hours.

Key Points:

Data is collected in batches.
Processing happens at a specific/scheduled time.
Suitable for large amounts of data.
It does not require real-time processing.

Simple example:
📥 Collect data → 📦 Make a batch → ⚙️ Process together → 📊 Get result

Streaming Data

Streaming Data is a method of processing data where data is processed continuously and immediately as it is generated or received, instead of waiting for a large amount of data to accumulate.

Example:
A food delivery app continuously receives the driver's GPS location and updates the customer's map in real time.

Key Points:

Data is processed continuously.
Processing happens in real time or near real time.
Suitable for continuously generated data.
Useful when quick results or immediate action is required.

Examples:

📍 GPS location tracking
💳 Online payment transactions
📱 Social media feeds
📈 Stock market prices
🌡️ IoT sensor data

Simple flow:
Data Generated → Process Immediately → Result

| Feature             | **Batch Processing**                                   | **Streaming Data Processing**                  |
| ------------------- | ------------------------------------------------------ | ---------------------------------------------- |
| **Meaning**         | Processes data in **groups/batches**                   | Processes data **continuously as it arrives**  |
| **Processing time** | At a **scheduled time**                                | **Immediately or near real-time**              |
| **Data size**       | Usually handles **large volumes** together             | Handles **continuous flow of data**            |
| **Result**          | Results are available **after the batch is processed** | Results are available **almost immediately**   |
| **Latency**         | **High latency**                                       | **Low latency**                                |
| **Suitable for**    | Tasks where immediate results are not required         | Tasks where immediate action is required       |
| **Example**         | Monthly salary/payroll processing                      | Live GPS tracking                              |
| **Other examples**  | Daily reports, billing, data backups                   | Stock prices, online transactions, IoT sensors |
| **Cost/complexity** | Generally **simpler** to implement                     | Generally **more complex** to implement        |


**6. Roles in the World of Data**

**The slide lists these major roles:**

Database Administrator (DBA) — Keeps databases secure, available, reliable, and performant.

Data Engineer — Builds data pipelines and infrastructure to collect, transform, and deliver data.

Data Visualization / Reporting professional — Turns data into reports, dashboards, and understandable insights.

Data Analyst — Interprets data to answer business questions and support decisions.

Data Scientist / ML professional — Uses statistical methods and machine-learning models to predict and uncover advanced insights.

| Role                    | Main Work                                    |
| ----------------------- | -------------------------------------------- |
| **DBA**                 | Manages and maintains databases              |
| **Data Engineer**       | Collects, cleans, transforms, and moves data |
| **Data Analyst**        | Analyzes data and finds insights             |
| **Data Visualization**  | Creates charts, reports, and dashboards      |
| **Data Scientist / ML** | Builds models and makes predictions          |
